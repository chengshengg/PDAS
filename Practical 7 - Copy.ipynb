    {'h': 1, 'e': 1, 'l': 3, 'o': 2, ' ': 1, 'w': 1, 'r': 1, 'd': 1}

## Question 4
Given a dictionary {'h': 1, 'e': 1, 'l': 3, 'o': 2, ' ': 1, 'w': 1, 'r': 1, 'd': 1}, find the single char that appears most frequently, and print it as the output. You cannot manually obeserve the the counts, but use a for loop to access the individual items. For example, the character l appears 3 times.

Expected Output: 
        
        The most frequent char:  l

```python
dict = {'h': 1, 'e': 1, 'l': 3, 'o': 2, ' ': 1, 'w': 1, 'r': 1, 'd': 1}
#write code below
highest = 0
character = ""

for key, value in dict.items():
    if value > highest:
        highest = value
        character = key

print("The most frequent char:", character)
```

## Question 6:
Complete the function find_special_numbers that finds all numbers between 1 and 100 (inclusive) that are divisible by both 3 and 5. 

Expected Output:

    Numbers divisible by both 3 and 5: [15, 30, 45, 60, 75, 90]
    Count of these numbers: 6

```python
def find_special_numbers():
    numbers = []
    for i in range(1, 101):
        if i % 3 == 0 and i % 5 == 0:
            numbers.append(i)
    
    count = len(numbers)
    return numbers, count

# Execute the function with fixed inputs
numbers, count = find_special_numbers()

#Do not modify the print statements
print("Numbers divisible by both 3 and 5:", numbers)
print("Count of these numbers:", count)
```

Finally, print the indices and the corresponding elements exactly as shown in the expected output.

Expected Output:

    Indices where arr > 5: (array([1, 3, 4], dtype=int64),)
