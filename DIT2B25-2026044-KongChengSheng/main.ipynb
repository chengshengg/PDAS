# Removing Irrelevant Columns

Irrelevant columns such as “Easter Egg” and “Unimportant Column” were removed because they do not contribute to answering the research question and contain large amounts of missing data.

Dataset 5 was not merged into the final dataframe because it contained duplicate information already present in Dataset 1. Including it would create redundant columns without contributing additional insights relevant to the research question.

```python
postings1 = postings1.loc[:, ~postings1.columns.str.contains('Easter Egg')]
postings1 = postings1.loc[:, ~postings1.columns.str.contains('Unimportant')]

postings5 = postings5.loc[:, ~postings5.columns.str.contains('Easter Egg')]
postings5 = postings5.loc[:, ~postings5.columns.str.contains('Unimportant')]
```

# Merging Datasets

The datasets are merged using the common job_id column to create a unified dataset for analysis.

```python
df = postings1.merge(postings2, on='job_id')
df = df.merge(postings3, on='job_id')
df = df.merge(postings4, on='job_id')
```

```python
#Inspecting the Merged Dataset
df.info()
df.head()
df.columns
df.isnull().sum()
```

# Data Cleaning

Data cleaning is performed to remove missing or invalid salary values so that the analysis is more reliable and meaningful.

Salary values in the dataset were recorded using different pay periods such as hourly, monthly, and yearly compensation. To ensure fair and meaningful salary comparisons, only yearly salary records were retained for analysis.

Rows with missing experience level values were also removed because experience level is another important variable in the research question.

```python
#Checking Pay Periods
df['pay_period'].value_counts()
```

# Cleaning Dataset

```python
# Remove rows with missing salary
df = df.dropna(subset=['max_salary'])

# Keep only positive salaries
df = df[df['max_salary'] > 0]

# Keep only yearly salaries
df = df[df['pay_period'] == 'YEARLY']

# Remove rows with missing experience level
df = df.dropna(subset=['formatted_experience_level'])
```

# Checking Cleaned Dataset

```python
print(df.shape)

df[['max_salary', 'pay_period', 'formatted_experience_level', 'location', 'work_type']].head()
```

# Exploratory Data Analysis Using Pandas

This section uses Pandas functions to identify trends and relationships between salary, experience level, work type, and location.

# Average Salary by Experience Level

```python
salary_by_experience = df.groupby('formatted_experience_level')['max_salary'].mean().sort_values()

print(salary_by_experience)
```

## Explanation

The groupby() function groups the dataset according to experience level, while the mean() function calculates the average salary for each category.

This helps answer the research question by showing how salaries differ according to experience level.

```python
salary_by_worktype = df.groupby('work_type')['max_salary'].mean().sort_values()

print(salary_by_worktype)
```

## Explanation

This analysis compares average salaries across different work types.

It helps determine whether certain work arrangements offer higher salaries.

```python
top_locations = df.groupby('location')['max_salary'].mean().sort_values(ascending=False).head(10)

print(top_locations)
```

## Explanation

This analysis identifies which locations offer the highest average salaries.

The analysis supports the research question by comparing salary differences across locations.

# Additional Pandas Method Research

The pivot_table() method was researched independently.

The pivot_table() method summarises and reorganises data into a more readable format.

This method helps reveal salary trends across multiple variables simultaneously.

```python
pivot_salary = pd.pivot_table(
    df,
    values='max_salary',
    index='formatted_experience_level',
    columns='work_type',
    aggfunc='mean'
)

print(pivot_salary)
```

## Interpretation

The pivot table shows how salary levels vary across both experience levels and work types simultaneously.

Higher experience levels generally receive higher salaries across most work arrangements, while full-time roles tend to have the highest average salaries overall.

# Data Visualization Using Matplotlib

## This section uses Matplotlib to create charts that visually represent patterns and relationships in the dataset.

# Chart 1 — Bar Chart

## Average Salary by Experience Level

```python
salary_by_experience.plot(kind='bar')

plt.title('Average Salary by Experience Level')
plt.xlabel('Experience Level')
plt.ylabel('Average Salary')

plt.xticks(rotation=45)

plt.show()
```

## Interpretation

The chart shows a clear positive relationship between experience level and yearly salary. Internship positions have the lowest average salary, while Director and Executive positions earn significantly higher salaries.

The largest salary increase occurs between Mid-Senior level and Director roles, suggesting that managerial and leadership positions receive substantially higher compensation.

This indicates that experience and seniority strongly influence salary levels.

# Chart 2 — Histogram
## Distribution of Yearly Salary

```python
average_salary = df['max_salary'].mean()

plt.hist(df['max_salary'], bins=30)

plt.axvline(
    average_salary,
    linestyle='dashed',
    linewidth=2,
    label='Average Salary'
)

plt.title('Distribution of Yearly Salaries')
plt.xlabel('Maximum Salary')
plt.ylabel('Frequency')

plt.ticklabel_format(style='plain', axis='x')

plt.legend()

plt.show()
```

## Interpretation

The histogram shows that most jobs offer salaries within the lower to middle salary ranges, while only a small number of jobs offer extremely high salaries.

This creates a long tail on the right side of the graph, indicating that high-paying jobs are less common in the dataset.

The dashed vertical line represents the average yearly salary. Most salary values fall below the average, suggesting that a small number of extremely high salaries increase the overall average salary.

# Chart 3 — Boxplot
## Salary Distribution by Work Type

```python
work_types = df['work_type'].unique()

salary_data = [
    df[df['work_type'] == work]['max_salary']
    for work in work_types
]

plt.boxplot(salary_data, tick_labels=work_types)

plt.title('Salary Distribution by Work Type')
plt.xlabel('Work Type')
plt.ylabel('Maximum Salary')

plt.xticks(rotation=45)

plt.ticklabel_format(style='plain', axis='y')

plt.show()
```

## Interpretation

The boxplot shows that FULL_TIME jobs have the widest salary distribution and the largest number of outliers, indicating a broad range of salaries for full-time positions.

CONTRACT jobs also contain several extremely high-paying outliers, suggesting that certain contract positions offer exceptionally high compensation.

PART_TIME and TEMPORARY jobs generally have lower median salaries and smaller salary ranges compared to FULL_TIME roles.

# Chart 4 — Pie Chart
## Distribution of Work Types

```python
worktype_counts = df['work_type'].value_counts()

# Combine small categories into OTHERS
small_categories = worktype_counts[worktype_counts < 100].sum()

worktype_counts = worktype_counts[worktype_counts >= 100]

worktype_counts['OTHERS'] = small_categories

plt.figure(figsize=(7,7))

plt.pie(
    worktype_counts,
    labels=worktype_counts.index,
    autopct='%1.1f%%'
)

plt.title('Distribution of Work Types')

plt.show()
```

## Interpretation

The pie chart shows that FULL_TIME positions dominate the dataset, accounting for the vast majority of job postings.

Smaller work arrangements such as PART_TIME, TEMPORARY, and VOLUNTEER positions were grouped into “OTHERS” to improve chart readability.

This suggests that employers primarily advertise full-time roles within the dataset.

# Chart 5 — Line Chart
## Average Salary by Top Locations

```python
plt.plot(top_locations.index, top_locations.values, marker='o')

plt.title('Average Salary by Top Locations')
plt.xlabel('Location')
plt.ylabel('Average Salary')

plt.xticks(rotation=45)

plt.show()
```

## Interpretation

The line chart shows that average yearly salaries vary significantly across locations.

Certain cities offer substantially higher salaries compared to others, indicating that geographic location plays an important role in salary differences.

Locations with higher salaries may reflect areas with stronger demand for specialized or senior-level jobs.

# Additional Matplotlib Feature Research

This function changes the formatting style of chart axis labels. It was used to disable scientific notation on the histogram x-axis, making salary values easier to read and interpret.

The `axvline()` function was also researched independently. This function adds a vertical reference line onto a chart. It was used to indicate the average yearly salary within the histogram, improving the readability and interpretability of the salary distribution.
