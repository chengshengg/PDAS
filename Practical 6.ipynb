### 1. Import Libraries and "fake" youtube dataset

This code creates a sample dataset using pandas, a popular Python library for data manipulation and analysis. The dataset represents a collection of video metrics, including views, likes, comments, shares, and upload hours.

Step-by-Step Breakdown:
1. Importing Libraries: Although not shown in this snippet, the code assumes that the necessary libraries, pandas (as pd) and numpy (as np), have been imported.
2. Creating a DataFrame: A pandas DataFrame named data is created with six columns:
- video_id: A unique identifier for each video, ranging from 1 to 100.
- views, likes, comments, shares: These columns represent video engagement metrics, populated with random integers within specified ranges to simulate real-world data.
- upload_hour: The hour of the day when the video was uploaded, represented by a random integer between 0 (midnight) and 23 (11 PM).

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Fake YouTube dataset of 100 data points
data = pd.DataFrame({
    "video_id": range(1, 101),
    "views": np.random.randint(1000, 100000, 100),
    "likes": np.random.randint(50, 5000, 100),
    "comments": np.random.randint(10, 1000, 100),
    "shares": np.random.randint(5, 500, 100),
    "upload_hour": np.random.randint(0, 24, 100)
})
print("Dataset preview:")
print(data.head())
```

## Remember that you've already mastered creating various plots in matplotlib. This notebook will show you how to seamlessly combine multiple plots to uncover deeper insights and create compelling visualizations.

## 2. Univariate Plots

This code generates a histogram with a kernel density estimate (KDE) to visualize the distribution of video views using seaborn and matplotlib.

Step-by-Step Breakdown:
1. Creating a Figure: plt.figure(figsize=(8, 5)) creates a new figure with a specified size (8 inches wide, 5 inches tall) for the plot.
2. Histogram with KDE: sns.histplot(data["views"], kde=True, bins=20) creates a histogram of the "views" column from the data DataFrame.
- kde=True adds a kernel density estimate (KDE) curve to the plot, which helps to smooth out the distribution and provide a better visualization of the underlying data.
- bins=20 specifies the number of bins for the histogram, which determines how the data is grouped and displayed.

```python
plt.figure(figsize=(8, 5))
sns.histplot(data["views"], kde=True, bins=20)
plt.title("Univariate Analysis: Views Distribution with KDE")
plt.show()
```

The histogram with a kernel density estimate (KDE) provides valuable insights into the distribution of video views:

1. Skewness: The distribution appears to be right-skewed, indicating that most videos have a relatively low number of views, while a few videos have an exceptionally high number of views.
2. Peaks and Modes: The KDE curve suggests a unimodal distribution, with a single peak around the median view count. This implies that most videos cluster around a typical view count.
3. Tails: The heavy tail on the right side of the distribution indicates the presence of outliers – videos with an unusually high number of views.
4. Central Tendency: The median view count (around the peak) provides a better representation of the typical view count than the mean, which would be skewed by the outliers.
5. Variability: The spread of the distribution indicates a moderate to high variability in view counts, suggesting that some videos perform significantly better than others.

## Here are different plots KDE plots, read them and understand their interpretation.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))  # 1 row, 2 columns of plots
# Plot KDE for 'views' on the first subplot
sns.kdeplot(data['views'], ax=axes[0], fill=True)
axes[0].set_title('KDE for Views')

# Plot KDE for 'likes' on the second subplot
sns.kdeplot(data['likes'], ax=axes[1], fill=True, color='red')
axes[1].set_title('KDE for Likes')
plt.tight_layout()  # Adjust layout for better spacing
plt.show()
```

## KDE for Views (First/Left Subplot)

This plot shows the distribution of video views. Here's what we can tell:

1. Shape: The distribution appears to be right-skewed, meaning there are more videos with lower view counts and a few videos with exceptionally high view counts. This is typical for online content, where a small percentage of videos go viral and attract a large audience.

2. Central Tendency: The peak of the KDE curve represents the mode or the most common view count. It's likely somewhere in the lower range of the x-axis, indicating that most videos have relatively fewer views.

3. Spread: The spread of the KDE curve shows the variability in view counts. The long tail on the right side indicates a wide range of view counts, with some videos having significantly more views than others.

## KDE for Likes (Second/Right Subplot)

This plot shows the distribution of video likes. Here's what we can tell:

1. Shape: Similar to views, the distribution of likes also appears to be right-skewed, indicating more videos with lower likes and a few with exceptionally high likes.

2. Central Tendency: The peak of the KDE curve represents the mode or the most common number of likes. It's likely lower than the peak for views, suggesting that videos generally receive fewer likes than views.

3. Spread: The spread of the KDE curve shows the variability in likes. It might be slightly narrower than the spread for views, indicating less variability in the number of likes compared to views.

## Comparison and Insights

### By comparing the two plots, we can draw some insights:

1. Relationship: Both views and likes exhibit a right-skewed distribution, suggesting a positive correlation between them. Videos with more views tend to have more likes.

2. Scale: The scale of the x-axis is different for the two plots, reflecting the difference in magnitude between views and likes. Views generally have higher values than likes.

3. Skewness: The skewness of both distributions highlights the presence of outliers – videos with exceptionally high views or likes. These outliers can significantly influence summary statistics like the mean.

3. Distribution: The KDE curves provide a visual representation of the probability density of views and likes. The higher the curve, the more likely it is to observe videos with those view or like counts.

### Overall Interpretation

The KDE plots suggest that video views and likes are positively correlated and exhibit a right-skewed distribution. This means most videos have relatively lower views and likes, while a small percentage of videos achieve high engagement. Understanding these distributions can be helpful for content creators and marketers in setting realistic expectations and optimizing their strategies.

Remember that these are just initial observations based on the KDE plots. Further analysis might be needed to explore the relationship between views and likes in more detail and to identify factors that influence video engagement.

## 3. Bivariate Analysis

Bivariate Analysis: Exploring Relationships Between Two Variables

Bivariate analysis is a statistical method that examines the relationship between two variables. It goes beyond simply describing individual variables and focuses on understanding how they interact or influence each other.

Key Goals of Bivariate Analysis

1. Identifying Relationships: It helps determine if a relationship exists between two variables and the nature of that relationship (e.g., positive, negative, linear, nonlinear).

2. Measuring Strength and Direction: It quantifies the strength and direction of the relationship. For example, a strong positive correlation indicates that as one variable increases, the other tends to increase as well.

3. Predicting Outcomes: By understanding the relationship between variables, we can potentially predict the value of one variable based on the value of the other.

4. Testing Hypotheses: Bivariate analysis can be used to test hypotheses about the relationship between variables. For instance, we might hypothesize that there's a positive correlation between advertising spending and sales revenue.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x="likes", y="views", hue="upload_hour", data=data)
plt.title("Bivariate Analysis: Likes vs Views")
plt.show()
```

Below is a created heatmap that visualizes the correlation between multiple variables (likes, views, comments, shares). It uses color intensity to represent the strength and direction of the correlations.

```python
correlation_matrix = data[["likes", "views", "comments", "shares"]].corr()
plt.figure(figsize=(8, 5))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

## 4. Multivariate Analysis

### Multivariate analysis, as the name suggests, involves exploring relationships between three or more variables simultaneously. It aims to understand the complex interactions and dependencies among multiple variables in a dataset.

Pair Plot: A Tool for Multivariate Visualization

The pairplot function in Seaborn is a powerful tool for visualizing multivariate data. It creates a grid of plots, where each plot represents the relationship between a pair of variables in the dataset. This allows you to quickly see patterns, correlations, and potential outliers across multiple variables.

sns.pairplot(data, hue="upload_hour"): This line creates the pair plot.

1. data: This is your Pandas DataFrame containing the data.
2. hue="upload_hour": This argument specifies a categorical variable ("upload_hour" in this case) to color the points in the plots based on their category. This helps visualize how the relationship between variables might differ across different categories.

```python
sns.pairplot(data, hue="upload_hour")
plt.show()
```

### Interpreting the Pair Plot

1. Diagonal Plots: The plots along the diagonal show the distribution of each individual variable using histograms or kernel density estimates (KDEs).
2. Off-Diagonal Plots: The plots off the diagonal show the relationship between pairs of variables using scatter plots. If the points cluster along a line, it indicates a correlation between the variables. The color of the points, based on the hue argument, can reveal how the relationship might change for different categories.

### Insights from the Pair Plot

By examining the pair plot, you can gain insights into:

1. Correlations: Identify potential correlations between variables by looking for patterns in the scatter plots.
Distributions: Understand the distribution of each individual variable from the diagonal plots.
2. Outliers: Detect potential outliers that might deviate from the general patterns.
3. Categorical Differences: Observe how the relationship between variables might differ across different categories based on the color of the points.

### Example Interpretation

If you see a strong positive correlation between "views" and "likes" in the scatter plot, it suggests that videos with more views tend to have more likes. If you notice different patterns in the scatter plot based on the "upload_hour" color, it might indicate that the relationship between views and likes varies depending on when the video was uploaded.

- Which branches perform best.

- How customer traffic influences revenue.

Without these insights, companies risk inefficient marketing spending and missed revenue opportunities.


# 1. Import and explore the dataset

```python
# Import the dataset
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("KFC_Past_Sales_Dataset.csv")

# Display the first few rows of the dataset
print("Dataset preview:")
print(df.head())
# Check for missing values in each column

print("\nDataset Information:")
print(df.info())
```

# 2. Cleaning the dataset

```python
# Check for missing values
print(df.isnull().sum())

# Check for duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Remove duplicate rows if any exist
df = df.drop_duplicates()

# Remove rows with missing values if any exist
df = df.dropna()

# Verify cleaned dataset
print(df.info())
```

##

# 3. Kernel Density Estimation Chart (Univariate Analysis)
This KDE plot shows the distribution of sales across all KFC branches and countries. It helps management identify the most common sales range and understand overall business performance.

```python
plt.figure(figsize=(10,6))

sns.kdeplot(
    data=df,
    x='Sales',
    fill=True
)

plt.title('Distribution of KFC Sales')
plt.xlabel('Sales')
plt.ylabel('Density')

plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

## Intepretation of the KDE Plot
The KDE plot visually represents the probability density of KFC sales across all branches and countries.

1. The peak of the curve indicates the most common sales range among KFC branches, which appears to be around $120,000 to $160,000.
2. The spread of the curve shows the variability in sales performance across branches. A wider distribution suggests that different branches generate significantly different levels of revenue.
3. The curve is slightly skewed towards the right, indicating that while most branches achieve moderate sales figures, a smaller number of branches generate exceptionally high sales.

This plot is useful for management to quickly understand the overall sales landscape and identify what constitutes typical branch performance.

# 4. Scatterplot
This scatter plot illustrates the relationship between marketing expenditure and sales performance.

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x='Marketing_Spend',
    y='Sales',
    alpha=0.4
)

plt.title('Marketing Spend vs Sales')
plt.xlabel('Marketing Spend')
plt.ylabel('Sales')

plt.grid(True, linestyle='--', alpha=0.6)



plt.show()
```

# Interpretation of Scatter Plot
The scatter plot shows individual data points where each point represents a branch with a specific marketing expenditure and corresponding sales value.

1. The points generally show an upward trend as marketing spend increases, suggesting a positive relationship between marketing investment and sales revenue.
2. Although higher marketing expenditure tends to be associated with higher sales, the spread of the points indicates that marketing spend is not the only factor affecting sales performance.
3. Branches with similar marketing budgets can still achieve different sales outcomes, implying that other factors such as location, customer preferences, or operational efficiency may also play important roles.
4. There are no major outliers in the dataset, suggesting relatively consistent business performance across branches.

This plot helps management understand whether marketing expenditure contributes to sales growth and assists in evaluating the effectiveness of marketing strategies.

# 5. Pair Plot
The scatter plot helps determine whether increased experience leads to higher salaries and identifies any outliers.

```python
pair_columns = [
    'Sales',
    'Customers',
    'Marketing_Spend'
]

sns.pairplot(df[pair_columns])

plt.show()
```

# Intepretation of the Pair Plot
The pair plot is a grid of scatter plots (off-diagonal) and distribution plots (diagonal).

## Diagonal: The plots on the diagonal show the individual distribution of each variable (Sales, Customers, and Marketing_Spend) using histograms. This provides a quick overview of the spread and concentration of each variable.
1. Sales Distribution: The sales distribution appears slightly right-skewed, indicating that most branches generate moderate sales while fewer branches achieve exceptionally high revenue.
2. Customers Distribution: The customer count appears relatively evenly distributed across the dataset, suggesting that branches experience a wide but balanced range of customer traffic.
3. Marketing Spend Distribution: Marketing expenditure is also fairly evenly distributed, indicating that branches receive varying levels of marketing support.

## Off-Diagonal (Scatter Plots)
1. The scatter plot between Customers and Sales shows a strong positive relationship, indicating that branches serving more customers generally generate higher sales revenue.
2. The scatter plot between Marketing Spend and Sales shows a moderate positive relationship, suggesting that increased marketing investment may contribute to higher sales performance.
3. The scatter plot between Marketing Spend and Customers suggests that marketing activities may help attract additional customers to KFC branches.

This pair plot allows management to explore the relationships between key business variables and gain a broader understanding of the factors influencing branch performance and revenue generation.

## 1. Import and Exploratory Data Analysis

```python
# Load the dataset
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
data_analyst_jobs = pd.read_csv('/content/drive/MyDrive/PDAS Ay2526/DataAnalyst.csv')

# Display the first few rows and information about the dataset
print("Dataset preview:")
print(data_analyst_jobs.head())
print("\nDataset Info:")
print(data_analyst_jobs.info())
```

# 2. Clean the data and create more columns

```python
# Clean column names by removing leading/trailing spaces
data_analyst_jobs.columns = data_analyst_jobs.columns.str.strip()

# Extract numerical salary range (assuming Salary Estimate is in a range like $XX-$YYK)
# This is a basic approach and might need refinement based on the actual format
data_analyst_jobs['Min Salary'] = data_analyst_jobs['Salary Estimate'].str.extract(r'\$(\d+)[Kk]').astype(float)
data_analyst_jobs['Max Salary'] = data_analyst_jobs['Salary Estimate'].str.extract(r'\$(\d+)[Kk]').astype(float)

# Calculate average salary
data_analyst_jobs['Avg Salary'] = (data_analyst_jobs['Min Salary'] + data_analyst_jobs['Max Salary']) / 2

# Drop rows with NaN in Avg Salary for plotting (as needed for some plots)
data_analyst_jobs_cleaned = data_analyst_jobs.dropna(subset=['Avg Salary', 'Rating', 'Size']) # Added 'Size' and 'Rating' for pairplot
```

 ### 1. Kernel Density Estimation Chart (Univariate Analysis)

 This KDE plot shows the distribution of average salaries for Data Analyst jobs. It helps job seekers understand the typical salary range they can expect and where the salaries are most concentrated.


```python
plt.figure(figsize=(10, 6))
sns.kdeplot(data=data_analyst_jobs_cleaned, x='Avg Salary', fill=True)
plt.title('Distribution of Average Salaries for Data Analyst Jobs (in K)')
plt.xlabel('Average Salary (K)')
plt.ylabel('Density')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

 **Interpretation of the KDE Plot:**

The KDE plot visually represents the probability density of average salaries.
 - The peak of the curve indicates the most common average salary range.
 - The spread of the curve shows the variability in salaries. A wider curve suggests a larger range of salaries, while a narrower curve indicates salaries are more clustered around the peak.
 - The shape of the curve (e.g., skewed to the right or left) reveals if there are more jobs with lower or higher average salaries.

 This plot is useful for job seekers to quickly grasp the general salary landscape and benchmark their expectations.


### 2. Scatter Plot (Bivariate Analysis)

 This scatter plot explores the relationship between Company Rating and Average Salary. Job seekers can use this to see if there's a correlation between how companies are rated and the salaries they offer.


```python
# Plot 2: Scatterplot of Rating vs Average Salary (Bivariate)
plt.figure(figsize=(10, 6))
sns.scatterplot(data=data_analyst_jobs_cleaned, x='Rating', y='Avg Salary', alpha=0.6)
plt.title('Company Rating vs Average Salary for Data Analyst Jobs')
plt.xlabel('Company Rating')
# Filter out rows where 'Avg Salary' is NaN before plotting
filtered_data = data_analyst_jobs_cleaned.dropna(subset=['Avg Salary'])

# Plot 2: Scatterplot of Rating vs Average Salary (Bivariate)
plt.figure(figsize=(10, 6))
sns.scatterplot(data=filtered_data, x='Rating', y='Avg Salary', alpha=0.6)
plt.title('Company Rating vs Average Salary for Data Analyst Jobs')
plt.xlabel('Company Rating')
plt.ylabel('Average Salary (K)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

**Interpretation of the Scatter Plot:**

 The scatter plot shows individual data points where each point represents a company with a specific rating and average salary.
- If the points tend to cluster upwards as the rating increases, it suggests a positive correlation (higher rated companies tend to offer higher salaries).
- If the points are scattered randomly, it indicates little or no linear relationship between rating and salary.
- Outliers (points far away from the main cluster) might represent companies with unusually high or low salaries for their rating.

This plot helps job seekers understand if company reputation (as reflected by rating) is a significant factor in salary expectations.


 ### 3. Pair Plot (Multivariate Analysis)

 This pair plot visualizes the relationships between multiple numerical variables: Company Rating, Minimum Salary, Maximum Salary, and Average Salary. It provides a comprehensive overview of how these factors interact.



```python
# Plot 3: Pairplot for Multivariate Analysis
# We'll include relevant numerical columns for pairwise relationships
numerical_cols = ['Rating', 'Min Salary', 'Max Salary', 'Avg Salary']

# Drop rows with NaN in any of the numerical_cols for the pairplot
pairplot_data = data_analyst_jobs_cleaned[numerical_cols].dropna()

sns.pairplot(pairplot_data)
plt.suptitle('Pairwise Relationships of Numerical Variables', y=1.02, fontsize=16) # Add a title to the pairplot
plt.show()
```

 **Interpretation of the Pair Plot:**

The pair plot is a grid of scatter plots (off-diagonal) and distribution plots (diagonal). - **Diagonal:** The plots on the diagonal show the univariate distribution of each variable (Rating, Min Salary, Max Salary, Avg Salary) using histograms or KDEs. This gives a quick overview of the distribution of each individual factor.
- **Off-Diagonal (Scatter Plots):** Each off-diagonal plot shows the bivariate relationship between two variables. For example, you can see the scatter plot of 'Min Salary' vs 'Max Salary', 'Rating' vs 'Min Salary', etc.
- By examining these scatter plots, you can identify potential correlations or patterns between pairs of variables. For instance, you would likely see a strong positive correlation between 'Min Salary' and 'Max Salary'. You can also look for patterns in the relationship between 'Rating' and the salary variables.

This pair plot allows job seekers to explore the interconnectedness of various factors that influence salary and gain a broader understanding of the data.
